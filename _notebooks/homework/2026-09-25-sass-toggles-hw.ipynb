## Nonprofit Application

PVO can use reusable toggle styles to let volunteers choose which outreach information is visible in a portal.

## Popcorn Hack: Toggle Components

The checkbox stores Boolean state; the track shows the visual switch; a label explains what is controlled.

```python
%%html
<!-- UI_RUNNER: PVO Toggle Popcorn | interactive: true -->
<label class="ocs__toggle ocs__toggle--green">
  <input class="ocs__toggle-input" type="checkbox" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show volunteer reminders</span>
</label>
```

## Homework Hack: PVO Volunteer Settings

Three labeled toggles control real interface elements. The enabled count updates when settings change.

```python
%%html
<!-- UI_RUNNER: PVO Toggle Settings | interactive: true -->
<div class="ocs__container"><div class="ocs__card" id="pvo-toggle-panel">
  <h2 class="ocs__section-title">PVO Volunteer Portal</h2>
  <p class="ocs__description">Choose which information you want to see.</p>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="pvo-reminders"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Reminders</span></label>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="pvo-shifts"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Upcoming shifts</span></label>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="pvo-tips"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Outreach tips</span></label>
  <div class="ocs__callout">Enabled: <output id="pvo-toggle-count" aria-live="polite">0</output></div>
  <p id="pvo-reminders" hidden>Remember to confirm your volunteer shift.</p>
  <p id="pvo-shifts" hidden>Check your assignment before the next event.</p>
  <p id="pvo-tips" hidden>Arrive a few minutes early to prepare supplies.</p>
</div></div>
<script>
(function () {
  const panel = document.getElementById('pvo-toggle-panel');
  const toggles = panel.querySelectorAll('.ocs__toggle-input');
  const update = function () {
    let total = 0;
    toggles.forEach(function (toggle) {
      document.getElementById(toggle.dataset.target).hidden = !toggle.checked;
      if (toggle.checked) total += 1;
    });
    document.getElementById('pvo-toggle-count').textContent = total;
  };
  toggles.forEach(function (toggle) { toggle.addEventListener('change', update); });
  update();
})();
</script>
```
