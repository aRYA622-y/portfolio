## Homework Hack: Interactive Practice Settings

Each OCS toggle controls a visible feature. The status count changes with the checkbox states.

```python
%%html
<!-- UI_RUNNER: toggles homework | interactive: true -->
<div class="ocs__container"><div class="ocs__card" id="arya-settings">
  <h2 class="ocs__section-title">Practice Settings</h2>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="arya-tip"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Show study tip</span></label>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="arya-plan"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Show practice plan</span></label>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="arya-reminder"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Show reminder</span></label>
  <p class="ocs__callout">Enabled: <output id="arya-enabled-count">0</output></p>
  <p id="arya-tip" hidden>Break large assignments into smaller steps.</p>
  <p id="arya-plan" hidden>Warm up, drill, then review.</p>
  <p id="arya-reminder" hidden>Return tools after the session.</p>
</div></div>
<script>
(function () {
 const panel = document.getElementById('arya-settings');
 const inputs = panel.querySelectorAll('.ocs__toggle-input');
 function refresh() {
   let enabled = 0;
   inputs.forEach(function (input) {
     document.getElementById(input.dataset.target).hidden = !input.checked;
     if (input.checked) enabled += 1;
   });
   document.getElementById('arya-enabled-count').textContent = enabled;
 }
 inputs.forEach(function (input) { input.addEventListener('change', refresh); });
 refresh();
})();
</script>
```

## Explanation

Each label contains a checkbox, track, and visible text. JavaScript reads checked state to update both the content and count.
