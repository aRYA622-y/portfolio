## Homework Hack: Page Volunteer Settings

Three labeled toggles control real interface elements. The enabled count updates when settings change.

```python
%%html
<!-- UI_RUNNER: Page Toggle Settings | interactive: true -->
<div class="ocs__container"><div class="ocs__card" id="page-toggle-panel">
  <h2 class="ocs__section-title">Page Volunteer Portal</h2>
  <p class="ocs__description">Choose which information you want to see.</p>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="page-reminders"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Reminders</span></label>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="page-shifts"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Upcoming shifts</span></label>
  <label class="ocs__toggle ocs__toggle--green"><input class="ocs__toggle-input" type="checkbox" data-target="page-tips"><span class="ocs__toggle-track" aria-hidden="true"></span><span class="ocs__toggle-label">Outreach tips</span></label>
  <div class="ocs__callout">Enabled: <output id="page-toggle-count" aria-live="polite">0</output></div>
  <p id="page-reminders" hidden>Remember to confirm your volunteer shift.</p>
  <p id="page-shifts" hidden>Check your assignment before the next event.</p>
  <p id="page-tips" hidden>Arrive a few minutes early to prepare supplies.</p>
</div></div>
<script>
(function () {
  const panel = document.getElementById('page-toggle-panel');
  const toggles = panel.querySelectorAll('.ocs__toggle-input');
  // .checked is the Boolean state; hidden is its inverse for each feature.
  // .checked is the Boolean state; hidden is its inverse for each feature.
  const update = function () {
    let total = 0;
    toggles.forEach(function (toggle) {
      document.getElementById(toggle.dataset.target).hidden = !toggle.checked;
      if (toggle.checked) total += 1;
    });
    // Recalculate from zero after every change so the counter never accumulates stale state.
    // Recalculate from zero after every change so the counter never accumulates stale state.
    document.getElementById('page-toggle-count').textContent = total;
  };
  toggles.forEach(function (toggle) { toggle.addEventListener('change', update); });
  update();
})();
</script>
```
