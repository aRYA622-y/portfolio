## Nonprofit Application

A community animal-rescue nonprofit can use semantic typography for accessible instructions on interacting gently with cats.

## Popcorn Hack: Semantic Text

Headings and a paragraph establish a clear hierarchy for the animal rescue's adoption advice.

```python
%%html
<!-- UI_RUNNER: Animal Rescue Typography | interactive: true -->
<h1 class="ocs__section-title">Community Animal Rescue</h1>
<h2>Meet the shelter cats</h2>
<p class="ocs__description">Give every cat time and space to get comfortable.</p>
```

## Homework Hack: How to Greet a Shelter Cat

Use headings, text, and an ordered list to explain a simple, respectful greeting.

```python
%%html
<!-- UI_RUNNER: Shelter Cat Instructions | interactive: true -->
<div class="ocs__card">
  <h2 class="ocs__section-title">How to Greet a Shelter Cat</h2>
  <h3>Start with a calm introduction</h3>
  <p class="ocs__description">Our rescue recommends allowing a cat to approach at its own pace.</p>
  <p><strong>Steps:</strong></p>
  <ol>
    <li>Ask a shelter volunteer whether the cat is comfortable meeting visitors.</li>
    <li>Hold out your hand and let the cat approach if it wants to.</li>
    <li>Gently pet the cat only if it seems comfortable.</li>
  </ol>
  <h3>Thank you for being kind to rescue animals!</h3>
</div>
```
