## Popcorn Hack: Text Hierarchy

Semantic headings and a paragraph present a robotics lesson clearly.

```python
%%html
<!-- UI_RUNNER: typography popcorn | interactive: true -->
<h1 class="ocs__section-title">NYX Robotics</h1>
<h2>Build Session</h2>
<p class="ocs__description">Practice safe and organized assembly.</p>
```

## Homework Hack: Organize a Build Station

Use a heading, descriptive text, and an ordered list to explain the procedure.

```python
%%html
<!-- UI_RUNNER: typography homework | interactive: true -->
<div class="ocs__card">
  <h2 class="ocs__section-title">How to Organize a Build Station</h2>
  <h3>Before beginning</h3>
  <p class="ocs__description">A tidy workspace makes it easier to find tools and check each part.</p>
  <p><strong>Follow these steps:</strong></p>
  <ol>
    <li>Clear the table and place tools in their labeled areas.</li>
    <li>Sort bolts, nuts, and spacers into separate containers.</li>
    <li>Check the parts list before beginning assembly.</li>
  </ol>
  <h3>You're ready to build!</h3>
</div>
```

## Explanation

Headings provide hierarchy; the paragraph gives context; the numbered list shows the procedure in order.
