## Popcorn Hack: Practice Scanner

Use a for loop to skip easy tasks and stop when a priority task appears.

```python
# CODE_RUNNER: popcorn - iteration
practice_scores = [2, 4, 1, 6, 9, 5]
for score in practice_scores:
    if score < 3:
        print("Skip:", score)
        continue
    print("Review:", score)
    if score >= 9:
        print("Priority score reached")
        break
```

## Homework Hack: Build Quality Scanner

Track processed, skipped, and flagged readings with counters, a running total, continue, and break.

```python
# CODE_RUNNER: homework - loop checker
measurements = [18, 45, 32, 61, 24, 78, 43, 96, 55]
safe_limit, critical_limit = 35, 90
processed = skipped = flagged = running_total = 0
critical_found = False
for value in measurements:
    processed += 1
    if value < safe_limit:
        skipped += 1
        continue
    flagged += 1
    running_total += value
    if value >= critical_limit:
        critical_found = True
        break
print("Processed:", processed, "Skipped:", skipped)
print("Flagged:", flagged, "Total:", running_total)
print("Critical:", critical_found)
assert (processed, skipped, flagged, running_total, critical_found) == (8, 3, 5, 323, True)
```

## Explanation

Continue skips only the current loop pass, while break exits the loop altogether. The report counts readings visited before the critical value.
