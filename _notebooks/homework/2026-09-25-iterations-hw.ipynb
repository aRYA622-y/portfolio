## Nonprofit Application

Safe Passage Heals can scan urgent relief requests using iteration, thresholds, counters, continue, and break.

## Popcorn Hack: Disaster Relief Alert Scanner

Skip routine alerts but stop scanning when a critical relief request appears.

```python
# CODE_RUNNER: Safe Passage Heals popcorn
relief_alerts = [1, 4, 2, 6, 3, 9, 5]
routine_limit, critical_limit = 3, 9
routine_count = action_count = action_total = 0
critical_found = False
for severity in relief_alerts:
    if severity < routine_limit:
        routine_count += 1
        print("Routine:", severity)
        continue
    action_count += 1
    action_total += severity
    print("Review:", severity)
    if severity >= critical_limit:
        critical_found = True
        break
print("Routine:", routine_count, "Action:", action_count, "Total:", action_total)
assert (routine_count, action_count, action_total, critical_found) == (2, 4, 22, True)
```

## Homework Hack: Relief Request Monitor

Scan nine priority scores, skipping routine requests while recording actionable requests; stop at the critical threshold.

```python
# CODE_RUNNER: Safe Passage Heals iterations homework
request_priorities = [22, 47, 35, 68, 51, 76, 44, 93, 59]
routine_limit, critical_limit = 40, 90
checked = skipped = action_count = total_priority = 0
critical_found = False
for priority in request_priorities:
    checked += 1
    if priority < routine_limit:
        skipped += 1
        continue
    action_count += 1
    total_priority += priority
    print("Needs attention:", priority)
    if priority >= critical_limit:
        critical_found = True
        break
print("Checked:", checked, "Skipped:", skipped, "Actions:", action_count)
print("Priority total:", total_priority, "Critical:", critical_found)
assert (checked, skipped, action_count, total_priority, critical_found) == (8, 2, 6, 379, True)
```

## Explanation

Continue skips the remaining statements for routine requests; break ends the scan after an urgent case. Accumulators summarize only the requests actually processed.
