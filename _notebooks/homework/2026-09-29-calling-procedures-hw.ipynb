## Nonprofit Application

PVO can reuse procedures to record volunteer outreach reports and identify requests requiring follow-up.

## Knowledge Check

A procedure is a named set of reusable instructions. A definition creates the procedure; a call executes it. Parameters are names in the definition, and arguments are the values provided at a call site.

## Popcorn Hack 1: Call a Procedure

Define one status-reporting procedure and call it with two different locations and activities.

```python
# CODE_RUNNER: PVO procedures popcorn 1
def report_status(location, activity):
    print(f"PVO update from {location}: {activity}")

report_status("Poway", "Food donation collection complete")
report_status("Rancho Bernardo", "Volunteer orientation scheduled")
```

## Popcorn Hack 2: Helper Procedure

A helper returns a Boolean to the reporting procedure; separate logic makes the outer procedure simpler.

```python
# CODE_RUNNER: PVO procedures popcorn 2
def needs_followup(status):
    return status == "Pending"

def process_report(location, status):
    if needs_followup(status):
        print("FOLLOW UP:", location, status)
    else:
        print("RECORDED:", location, status)

process_report("Poway", "Complete")
process_report("Rancho Bernardo", "Pending")
```

## Homework Hack: Volunteer Support Dispatcher

A helper classifies support requests while the outer procedure uses parameters to report an appropriate next step.

```python
# CODE_RUNNER: PVO procedures homework
def requires_followup(request_type):
    return request_type in ("Missing Supplies", "Unfilled Shift")

def dispatch_support(location, request_type):
    if requires_followup(request_type):
        result = f"FOLLOW-UP REQUIRED: {request_type} at {location}"
    else:
        result = f"STANDARD UPDATE: {request_type} at {location}"
    print(result)
    return result

assert dispatch_support("Poway", "Missing Supplies").startswith("FOLLOW-UP REQUIRED")
assert dispatch_support("Rancho Bernardo", "Volunteer Check-In").startswith("STANDARD UPDATE")
assert dispatch_support("Poway", "Unfilled Shift").startswith("FOLLOW-UP REQUIRED")
```

## Explanation

The helper stores the follow-up rule in one place; using location and request parameters allows one dispatcher to process many nonprofit requests.
