## Popcorn Hack 2

```python
# CODE_RUNNER: Emergency report helper
def is_priority(incident_type):
    return incident_type == "Medical Emergency"
def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print(f"PRIORITY: {incident_type} reported in {location}")
    else:
        print(f"STANDARD REPORT: {incident_type} reported in {location}")
process_report("Poway", "Medical Emergency")
process_report("Rancho Bernardo", "Power Outage")
```

## Homework

```python
# CODE_RUNNER: Emergency response dispatcher
def check_priority(incident_type):
    return incident_type == "Medical Emergency" or incident_type == "Fire"
def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)
    if priority:
        print(f"PRIORITY RESPONSE: {incident_type} at {location}")
    else:
        print(f"STANDARD RESPONSE: {incident_type} at {location}")
dispatch_response("Poway", "Medical Emergency")
dispatch_response("Rancho Bernardo", "Power Outage")
dispatch_response("Poway", "Fire")
```

## Explanation
The Boolean helper keeps the priority rule in one place. The outer procedure reuses it for three incident types and two locations, so changing the rule does not require editing every call.
