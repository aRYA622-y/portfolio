## Popcorn Hack 1: Parameters and Arguments

Define a procedure once, then call it with different inputs.

```python
# CODE_RUNNER: popcorn 1
def report_test(robot, result):
    print(f"{robot}: {result}")
report_test("NYX Prototype", "Drive test passed")
report_test("Practice Bot", "Intake needs adjustment")
```

## Popcorn Hack 2: Helper Procedure

A helper makes one decision that a second procedure can reuse.

```python
# CODE_RUNNER: popcorn 2
def is_priority(result):
    return result == "Failed"
def process_test(component, result):
    if is_priority(result):
        print("Priority fix:", component)
    else:
        print("Log result:", component)
process_test("Drivetrain", "Passed")
process_test("Intake", "Failed")
```

## Homework Hack: Robot Test Reporter

Use a helper returning a Boolean and a main procedure with two parameters; test priority and ordinary results.

```python
# CODE_RUNNER: homework - calling procedures
def requires_repair(status):
    return status in ("Failed", "Unsafe")

def dispatch_repair(component, status):
    if requires_repair(status):
        response = f"REPAIR NEEDED: {component} ({status})"
    else:
        response = f"TEST LOGGED: {component} ({status})"
    print(response)
    return response

assert dispatch_repair("Drivetrain", "Passed").startswith("TEST LOGGED")
assert dispatch_repair("Intake", "Failed").startswith("REPAIR NEEDED")
assert dispatch_repair("Arm", "Unsafe").startswith("REPAIR NEEDED")
```

## Explanation

Parameters make the main procedure reusable. The helper centralizes the repair rule, reducing duplicated logic.
