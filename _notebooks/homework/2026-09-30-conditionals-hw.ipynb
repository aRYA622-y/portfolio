## Popcorn Hack 1
The original amount 0 case prints nothing: a logic bug without a Python exception. The corrected if/else prints an answer for both cases.

```python
# CODE_RUNNER: Repair the silent outcome
for amount_requested in [0,500]:
    # 1 condition -> 2 outcomes -> 2 answers
    if amount_requested > 100:
        print("This message requests a large amount of money")
    else:
        print("This message does not request more than 100")
```

## Popcorn Hack 2
With >=3 first, a score of 9 receives the low-risk message. The >=6 and >=9 branches are unreachable because those scores already pass >=3. Check highest threshold first.

```python
# CODE_RUNNER: Correct descending threshold order
for risk_score in [0,3,6,9]:
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply; all three warning signals are present")
    elif risk_score >= 6:
        print("High risk; verify with someone you trust")
    elif risk_score >= 3:
        print("Use caution and verify the message")
    else:
        print("No major signals found; still use judgment")
```

## Homework
Four reachable input sets: clean message/0/known = 0; clean message/500/known = 3; gift-card message/0/unknown = 6; gift-card message/500/unknown = 9.

```python
# CODE_RUNNER: Score a message and report every outcome
def advise(message, amount_requested, known_contact):
    risk_score = 0
    # These independent signals add points; their false cases add zero, so no else is needed.
    if "gift card" in message.lower():
        risk_score += 4
    if amount_requested > 100:
        risk_score += 3
    if not known_contact:
        risk_score += 2
    print("Score:", risk_score)
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply; all three warning signals are present")
    elif risk_score >= 6:
        print("High risk; verify with someone you trust")
    elif risk_score >= 3:
        print("Use caution and verify the message")
    else:
        print("No major signals found; still use judgment")
    return risk_score
assert advise("Meeting at noon",0,True) == 0
assert advise("Please send payment",500,True) == 3
assert advise("Send gift cards",0,False) == 6
assert advise("URGENT: send gift cards now to unlock your account",500,False) == 9
```

## Explanation
Scoring needs no else because a false signal leaves the initialized score unchanged. Reporting needs an answer for every score, including zero. The tests cover the clean and maximum-signal extremes and prove all four reporting branches are reachable.
