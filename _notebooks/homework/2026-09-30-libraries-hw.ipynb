## Popcorn Hack 2
Changed the roster and added /coach above test_client. Flask supplies routing and JSON conversion.

````python
# CODE_RUNNER: Tennis scoreboard routes
from flask import Flask, jsonify
app = Flask(__name__)
roster = ["Practice Player", "Practice Partner"]
@app.route("/score")
def score():
    return jsonify({"games_won":12})
@app.route("/roster")
def team_roster():
    return jsonify({"players":roster})
@app.route("/coach")
def coach():
    return jsonify({"coach":"Demo Coach", "focus":"serve placement"})
with app.test_client() as client:
    for route in ["/score","/roster","/coach"]:
        result = client.get(route)
        print(route, result.status_code, result.get_json())
        assert result.status_code == 200
````

## Homework

````python
# CODE_RUNNER: Tennis season summary
import json
import datetime
import statistics
def summarize_season(games):
    """Parameters: games, a nonempty list of dictionaries with opponent and points (games won).
    Returns: a dictionary containing total_points, average_points, best_game, and points_by_opponent.
    For an empty list, returns zero totals, best_game None, and an empty opponent mapping.
    """
    if not games:
        return {"total_points":0,"average_points":0,"best_game":None,"points_by_opponent":{}}
    points = [game["points"] for game in games]
    by_opponent = {}
    for game in games:
        name = game["opponent"]
        by_opponent[name] = by_opponent.get(name,0) + game["points"]
    return {"total_points":sum(points),"average_points":statistics.mean(points),
            "best_game":max(games,key=lambda game:game["points"]),"points_by_opponent":by_opponent}
games = [{"opponent":"North Team","points":8},{"opponent":"West Team","points":12},
         {"opponent":"North Team","points":10},{"opponent":"South Team","points":9}]
report = summarize_season(games)
# Fixed analysis date makes the assignment's output reproducible.
report_date = datetime.date(2026,9,30)
next_game = datetime.date(2026,10,8)
report["days_until_next_game"] = (next_game - report_date).days
print(json.dumps(report,indent=2))
assert report["total_points"] == 39 and report["average_points"] == 9.75
assert report["days_until_next_game"] == 8
assert report["points_by_opponent"]["North Team"] == 18
print("Empty season:", summarize_season([]))
````
