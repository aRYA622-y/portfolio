## Popcorn Hack 1: Inclusive Random Integers

Randint includes both endpoints; repeated outputs are possible.

```python
# CODE_RUNNER: popcorn 1
import random
for attempt in range(5):
    print(random.randint(1, 6))
```

## Popcorn Hack 2: Select a Whole Record

Choosing a full dictionary keeps associated fields together.

```python
# CODE_RUNNER: popcorn 2
import random
parts = [{"name": "Wheel", "bin": "A1"}, {"name": "Motor", "bin": "B2"}, {"name": "Sensor", "bin": "C3"}]
selected = random.choice(parts)
print(selected["name"], selected["bin"])
```

## Popcorn Hack 3: Random Action

Choose a part and a test action independently, then branch on the action.

```python
# CODE_RUNNER: popcorn 3
import random
part = random.choice(["Wheel", "Motor", "Sensor"])
action = random.choice(["inspect", "test", "record"])
if action == "inspect":
    print("Inspecting", part)
elif action == "test":
    print("Testing", part)
else:
    print("Recording", part)
```

## Homework Hack: Parts Inventory QA Simulator

Run five randomized record/action checks. Test duplicate IDs before accepting a new record.

```python
# CODE_RUNNER: homework - random values
import random
inventory = [{"name": "Wheel", "id": "W1"}, {"name": "Motor", "id": "M2"}, {"name": "Sensor", "id": "S3"}]
existing_ids = {"W1", "S3"}
actions = ["search", "create", "update", "remove"]
for test_number in range(1, 6):
    item = random.choice(inventory)
    action = random.choice(actions)
    if action == "create":
        outcome = "duplicate" if item["id"] in existing_ids else "accepted"
    elif action == "search":
        outcome = "found"
    elif action == "update":
        outcome = "update simulated"
    else:
        outcome = "remove simulated"
    print(f"Test {test_number}: {action} {item['name']} -> {outcome}")
```

## Explanation

Random selection is inclusive for randint, and individual choices may repeat. These are simulated QA actions; the program does not modify a real inventory.
