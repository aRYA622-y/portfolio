## Popcorn Hack 3: Randomized API Path

Select a record and a simulated API action, then use a conditional to choose the corresponding result.

```python
# CODE_RUNNER: SFI random popcorn 3
import random
record = random.choice([
 {"product_name":"Racing Flywheel","spec_number":"1.1"},
 {"product_name":"Clutch Assembly","spec_number":"1.2"}])
action = random.choice(["GET/search","POST/create","PUT/update"])
existing = {"1.1"}
if action == "GET/search":
    print("SEARCH:", record["product_name"])
elif action == "POST/create":
    print("CREATE:", "duplicate" if record["spec_number"] in existing else "accepted")
else:
    print("UPDATE preview:", record["spec_number"])
```

## Homework Hack: SFI Backend QA Simulator

Run five randomized simulations using structured records and four actions. Validate duplicate specifications before allowing simulated creation.

```python
# CODE_RUNNER: SFI Foundation random QA homework
import random
records = [
 {"product_name":"Racing Flywheel","category":"Auto Racing","spec_number":"1.1"},
 {"product_name":"Clutch Assembly","category":"Drag Racing","spec_number":"1.2"},
 {"product_name":"Safety Harness","category":"Auto Racing","spec_number":"2.3"}]
actions = ["GET/search","POST/create","PUT/update","DELETE/remove"]
existing_specs = {"1.1", "2.3"}
def simulate_qa(record=None, action=None):
    record = random.choice(records) if record is None else record
    action = random.choice(actions) if action is None else action
    if action == "GET/search":
        outcome = "found"
    elif action == "POST/create":
        outcome = "duplicate" if record["spec_number"] in existing_specs else "accepted"
    elif action == "PUT/update":
        outcome = "update preview"
    elif action == "DELETE/remove":
        outcome = "remove preview"
    else:
        raise ValueError("Unknown action")
    return {"action": action, "spec_number": record["spec_number"], "outcome": outcome}
for number in range(1, 6):
    print("Test", number, simulate_qa())
assert simulate_qa(records[0], "POST/create")["outcome"] == "duplicate"
assert simulate_qa(records[1], "POST/create")["outcome"] == "accepted"
for action in actions:
    assert simulate_qa(records[0], action)["action"] == action
print("All action paths checked deterministically")
```

## Explanation

These are simulated backend actions and do not change real records. Multiple runs demonstrate several possible paths but do not guarantee every combination.
