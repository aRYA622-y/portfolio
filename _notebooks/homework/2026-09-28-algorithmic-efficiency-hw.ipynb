## Popcorn
Prediction for 1,000 books: linear worst case 1,000 checks, binary worst case 10. Binary search assumes sorted entries and direct access to a middle entry.

```python
# CODE_RUNNER: Catalog search comparisons
for n in [8,1000]:
    linear_checks = n
    remaining, binary_checks = n, 0
    while remaining > 0:
        binary_checks += 1
        remaining //= 2
    print("Books:", n, "Linear:", linear_checks, "Binary:", binary_checks)
```

## MCQ
Answers: A, B, B, C. Binary search needs sorted direct-access entries; compare growing work and memory; quadratic work quadruples when n doubles; heuristics do not guarantee an optimum. Answer review: 4/4.

## Homework
Predictions: 12/12/144, 24/24/576, and 0/0/0.

```python
# CODE_RUNNER: Library work models
for n in [12,24,0]:
    visits = n
    ordered_pairs = n * n
    print("Books:", n)
    print("Visits:", visits)
    print("Ordered pairs:", ordered_pairs)
    assert visits == n and ordered_pairs == n ** 2
```

## Tests and Recommendation
Actual outputs are recorded in the code cells. Doubling 12 to 24 doubles visits but quadruples ordered pairs. The empty catalog requires zero work in both models. For repeated searches of a sorted 1,000-book catalog, binary search reduces worst-case checks from 1,000 to 10; a target near the front can still favor linear search in one particular run. An unsorted catalog first needs sorting for binary search, and an extra sorted copy uses memory.

## Design Thinking
A librarian needs fast repeated lookups as the catalog grows. Linear and binary search were compared with a small calculator that labels the book count and work. The zero-book test confirmed both counts remain zero. A short-books-first reading heuristic may be useful, but it does not guarantee the reader’s best possible selection.
