## Popcorn Hack 1: Trace the Searches

For a sorted list, count linear and binary search comparisons with the same target.

```python
# CODE_RUNNER: popcorn 1
data = [3, 7, 11, 15, 19, 23, 27]
target = 19
linear_checks = 0
for item in data:
    linear_checks += 1
    if item == target:
        break
left, right, binary_checks = 0, len(data) - 1, 0
while left <= right:
    middle = (left + right) // 2
    binary_checks += 1
    if data[middle] == target:
        break
    if data[middle] < target:
        left = middle + 1
    else:
        right = middle - 1
print(linear_checks, binary_checks)
```

## Popcorn Hack 2: Compare Growth

Compare worst-case checks for linear and binary search at different input sizes.

```python
# CODE_RUNNER: popcorn 2
import math
for n in [8, 64, 512]:
    print(n, "linear:", n, "binary:", math.floor(math.log2(n)) + 1)
```

## Popcorn Hack 4: Tradeoff

Binary search requires sorted input. Linear search works with unsorted lists without sorting them first.

## Homework Hack: Compare Search Approaches

For three input sizes, calculate worst-case comparisons and describe when sorted input makes binary search useful.

```python
# CODE_RUNNER: homework - algorithmic efficiency
import math
sizes = [20, 200, 2000]
comparison = []
for n in sizes:
    linear = n
    binary = math.floor(math.log2(n)) + 1
    comparison.append((n, linear, binary))
    print(f"{n} records: linear <= {linear} checks; binary <= {binary} checks")
assert comparison == [(20, 20, 5), (200, 200, 8), (2000, 2000, 11)]
```

## Explanation

Binary search reduces comparisons on a sorted list, but preparing an unsorted list may cost more than one simple linear search. Input organization matters.
