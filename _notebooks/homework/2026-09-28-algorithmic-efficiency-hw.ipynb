## Nonprofit Application

A nonprofit directory of community resources can compare linear and binary search when looking up sorted resource IDs.

## Popcorn Hack 1: Compare Searches

Trace both searches on one sorted directory. A target at the beginning can favor linear search on that particular run.

```python
# CODE_RUNNER: nonprofit resource lookup popcorn 1
resource_ids = [3, 6, 9, 12, 15, 18, 21]
target = 3
linear_checks = 0
for value in resource_ids:
    linear_checks += 1
    if value == target:
        break
left, right, binary_checks = 0, len(resource_ids) - 1, 0
while left <= right:
    middle = (left + right) // 2
    binary_checks += 1
    if resource_ids[middle] == target:
        break
    if resource_ids[middle] < target:
        left = middle + 1
    else:
        right = middle - 1
print("Linear:", linear_checks, "Binary:", binary_checks)
assert (linear_checks, binary_checks) == (1, 3)
```

## Popcorn Hack 2: Compare Growth

Linear worst-case comparisons scale with the entire list; binary search grows logarithmically.

```python
# CODE_RUNNER: efficiency popcorn 2
import math
for size in [1000, 2000]:
    print(size, "linear:", size, "binary:", math.floor(math.log2(size)) + 1)
```

## Popcorn Hack 3: Operation Counts

Compare linear, quadratic, and exponential estimates for two input sizes.

```python
# CODE_RUNNER: efficiency popcorn 3
for size in [10, 20]:
    print(size, "linear:", size, "quadratic:", size ** 2, "exponential:", 2 ** size)
```

## Popcorn Hack 4: Real-World Tradeoff

A nonprofit driver may choose the closest next delivery stop as a fast heuristic; that choice does not guarantee the shortest possible route.

## Homework Hack: Search Algorithm Choice

For three already-sorted directory sizes, calculate linear and binary worst-case checks, then explain the preparation and memory tradeoffs.

```python
# CODE_RUNNER: nonprofit directory efficiency homework
import math
comparisons = []
for size in [100, 1000, 10000]:
    linear_checks = size
    binary_checks = math.floor(math.log2(size)) + 1
    comparisons.append((size, linear_checks, binary_checks))
    print(f"{size} resources: linear {linear_checks}, binary {binary_checks}")
assert comparisons == [(100, 100, 7), (1000, 1000, 10), (10000, 10000, 14)]
```

## Recommendation and Tradeoffs

For repeated searches in an already sorted directory, binary search uses fewer worst-case comparisons. For one search in an unsorted directory, linear search avoids sorting overhead. Retaining an extra sorted copy consumes memory.
