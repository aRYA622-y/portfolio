## Popcorn Hack 1: Check Required Fields

Each comparison creates a Boolean. Membership checks whether a value is in an allowed collection.

```python
# CODE_RUNNER: popcorn 1
member = {"name": "Arya", "group": "Build", "id": "N17", "active": True}
allowed_groups = ["Build", "Programming"]
print("Name:", member["name"].strip() != "")
print("Group:", member["group"] in allowed_groups)
print("ID:", member["id"].strip() != "")
print("Active:", member["active"])
```

## Popcorn Hack 2: Prevent Duplicates

Use AND and NOT so every requirement is met and the ID is not already registered.

```python
# CODE_RUNNER: popcorn 2
registered_ids = ["N10", "N11"]
def can_register(member):
    return (member["name"].strip() != "" and member["group"] in ["Build", "Programming"]
            and member["id"].strip() != "" and member["active"]
            and member["id"] not in registered_ids)
print(can_register({"name": "Arya", "group": "Build", "id": "N17", "active": True}))
print(can_register({"name": "Arya", "group": "Build", "id": "N10", "active": True}))
```

## Popcorn Hack 3: Search Conditions

OR allows two possible query matches; AND requires the record to be active.

```python
# CODE_RUNNER: popcorn 3
records = [{"name": "Arya", "id": "N17", "active": True}, {"name": "Kai", "id": "N18", "active": False}]
query = "N17"
found = [r for r in records if (query == r["name"] or query == r["id"]) and r["active"]]
print(found)
```

## Homework Hack: Club Registration Validator

Validate a good record, a duplicate, and an unsupported group using one reusable Boolean function.

```python
# CODE_RUNNER: homework - Boolean expressions
existing_ids = {"N10", "N11"}
def validate(record):
    has_name = bool(record["name"].strip())
    valid_group = record["group"] in ("Build", "Programming")
    has_id = bool(record["id"].strip())
    unique_id = record["id"] not in existing_ids
    return has_name and valid_group and has_id and unique_id

cases = [{"name": "Arya", "group": "Build", "id": "N17"},
         {"name": "Alex", "group": "Build", "id": "N10"},
         {"name": "Sam", "group": "Art", "id": "N19"}]
results = [validate(item) for item in cases]
print(results)
assert results == [True, False, False]
```

## Explanation

AND requires all fields, OR allows either search match, and NOT can invert a duplicate check. The tests cover accepted and rejected records.
