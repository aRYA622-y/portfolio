## Popcorn Hack 1: Record Checks

Evaluate each required field separately and print its Boolean result.

Prediction: all four field checks are True. Creation rejects the duplicate, then accepts the new number. Search keeps only the accepted-category record.

```python
# CODE_RUNNER: SFI Foundation popcorn 1
record = {"product_name":"Lightweight Racing Clutch", "category":"Drag Racing", "spec_number":"1.4", "effective_date":"2026-07-12"}
accepted_categories = ["Auto Racing", "Drag Racing"]
name_valid = record["product_name"].strip() != ""
category_valid = record["category"] in accepted_categories
spec_valid = record["spec_number"].strip() != ""
date_valid = record["effective_date"].strip() != ""
print(name_valid, category_valid, spec_valid, date_valid)
assert all([name_valid, category_valid, spec_valid, date_valid])
```

## Popcorn Hack 2: Record Creation Rule

Combine AND with NOT to reject records that have missing fields or duplicate specifications.

```python
# CODE_RUNNER: SFI Foundation popcorn 2
existing_spec_numbers = {"1.1", "1.3"}
def can_create(record):
    required = (bool(record["product_name"].strip())
                and record["category"] in ["Auto Racing", "Drag Racing"]
                and bool(record["spec_number"].strip())
                and bool(record["effective_date"].strip()))
    return required and record["spec_number"] not in existing_spec_numbers

record = {"product_name":"Racing Clutch", "category":"Drag Racing", "spec_number":"1.3", "effective_date":"2026-07-12"}
print("Duplicate accepted:", can_create(record))
record["spec_number"] = "1.4"
print("New record accepted:", can_create(record))
assert can_create(record)
```

## Popcorn Hack 3: Search Filter

Use OR for product-name or specification matches and AND for an accepted category.

```python
# CODE_RUNNER: SFI Foundation popcorn 3
query = "1.4"
records = [
 {"product_name":"Lightweight Racing Clutch", "category":"Drag Racing", "spec_number":"1.4"},
 {"product_name":"Racing Flywheel", "category":"Auto Racing", "spec_number":"2.3"},
 {"product_name":"Street Clutch", "category":"Street Car", "spec_number":"1.4"}
]
matches = [r for r in records
           if (query == r["product_name"] or query == r["spec_number"])
           and r["category"] in ("Auto Racing", "Drag Racing")]
print(matches)
assert len(matches) == 1
```

## Homework Hack: SFI Record Validator

Run a reusable validator against accepted, invalid-category, and duplicate test records.

```python
# CODE_RUNNER: SFI nonprofit homework
existing_specs = {"1.1", "2.1"}
def valid_sfi_record(r):
    name_ok = bool(r["product_name"].strip())
    category_ok = r["category"] in ("Auto Racing", "Drag Racing")
    number_ok = bool(r["spec_number"].strip())
    date_ok = bool(r["effective_date"].strip())
    not_duplicate = r["spec_number"] not in existing_specs
    return name_ok and category_ok and number_ok and date_ok and not_duplicate

samples = [
 {"product_name":"Safety Harness", "category":"Auto Racing", "spec_number":"2.4", "effective_date":"2026-09-29"},
 {"product_name":"Safety Harness", "category":"Other", "spec_number":"2.5", "effective_date":"2026-09-29"},
 {"product_name":"Safety Harness", "category":"Drag Racing", "spec_number":"1.1", "effective_date":"2026-09-29"}
]
results = [valid_sfi_record(r) for r in samples]
print(results)
assert results == [True, False, False]
```

## Explanation

All required conditions must pass, so the validator uses AND. It excludes duplicates using NOT and permits either of two categories using membership.

```python
# CODE_RUNNER: Missing-field validator tests
base = samples[0].copy()
for field in ["product_name", "spec_number", "effective_date"]:
    missing = base.copy()
    missing[field] = ""
    assert not valid_sfi_record(missing)
print("Missing-field tests passed")
```
