## SASS Buttons for MyGoodBrain

The nonprofit's wellness-resource links, check-in controls, and community actions are styled using the OCS SASS button grammar. All examples connect the button concepts to MyGoodBrain.

## Popcorn Hack 1: Add the Base Button Class

Use `ocs__btn` to style the MyGoodBrain resources link.

```python
%%html
<!-- UI_RUNNER: Popcorn 1 — MyGoodBrain Resource Link | interactive: true -->
<a class="ocs__btn" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener noreferrer">View MyGoodBrain Resources</a>
```

## Popcorn Hack 2: Modifiers and Button Groups

Use `pill`, `fill`, `outline`, and `ocs__links` to group a prominent daily check-in and a secondary resource link.

```python
%%html
<!-- UI_RUNNER: Popcorn 2 — MyGoodBrain Button Group | interactive: true -->
<div class="ocs__links">
  <button type="button" class="ocs__btn accent pill fill" onclick="document.getElementById('mgb-popcorn-status').textContent='Daily check-in opened (demo).'">Start Daily Check-In</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener noreferrer">Browse Wellness Resources</a>
</div>
<p id="mgb-popcorn-status" role="status" aria-live="polite">Select an action.</p>
```

## Popcorn Hack 3: Status Tones and Wide Layout

The `ocs__links--wide` layout organizes self-care and support resource links with distinct status tones.

```python
%%html
<!-- UI_RUNNER: Popcorn 3 — Wide Action Bar | interactive: true -->
<div class="ocs__links--wide">
  <a class="ocs__btn alert-green outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener noreferrer">Self-Care Resources</a>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener noreferrer">Crisis Hotline</a>
</div>
```

## Homework Hack 2: Action Priority

A filled check-in button receives primary emphasis. Secondary actions use outline styling; the demo buttons provide visible feedback and do not store personal information.

```python
%%html
<!-- UI_RUNNER: Homework 2 — MyGoodBrain Check-In Actions | interactive: true -->
<div class="ocs__card" id="mgb-action-panel">
  <h2 class="ocs__section-title">MyGoodBrain Daily Check-In</h2>
  <div class="ocs__links">
    <button type="button" class="ocs__btn accent fill" id="mgb-entry">Log Mood Entry</button>
    <button type="button" class="ocs__btn outline" id="mgb-trends">View Weekly Trends</button>
    <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener noreferrer">Crisis Hotline</a>
  </div>
  <p id="mgb-feedback" role="status" aria-live="polite">Choose an action.</p>
</div>
<script>
(function () {
  const feedback = document.getElementById('mgb-feedback');
  document.getElementById('mgb-entry').addEventListener('click', function () {
    feedback.textContent = 'Check-in demo opened. A connected application would save the entry.';
  });
  document.getElementById('mgb-trends').addEventListener('click', function () {
    feedback.textContent = 'Weekly trend preview selected.';
  });
})();
</script>
```

## Explanation

The `ocs__btn` base class supplies styling, while `fill`, `outline`, `pill`, and alert colors express action hierarchy. The OCS link groups handle layout without custom CSS. Buttons use click handlers and live feedback, so the actions are functional.
