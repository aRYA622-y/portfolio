## SASS Buttons

These examples use the existing OCS SASS button classes. The popcorn hacks practice the button grammar, and the homework hacks create usable robotics controls.

## Popcorn Hack 1: Base Button

The `ocs__btn` class applies shared button styling. The anchor navigates to the robotics resources section in this notebook.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 1 — Base Button | interactive: true -->
<a class="ocs__btn" href="#arya-button-resources">View Robotics Resources</a>
<section id="arya-button-resources">
  <p>Robotics resources: build checklist, team notes, and practice schedule.</p>
</section>
```

## Popcorn Hack 2: Modifiers and Button Groups

Group related actions with `ocs__links`. The filled pill emphasizes the primary action, while the outlined link is secondary.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 2 — Button Modifiers | interactive: true -->
<div class="ocs__links">
  <button type="button" class="ocs__btn accent pill fill" onclick="this.nextElementSibling.textContent='Build checklist opened: inspect, assemble, test.'">Open Build Checklist</button>
  <output aria-live="polite">Select an action.</output>
  <a class="ocs__btn outline" href="#arya-practice-notes">Practice Notes</a>
</div>
<p id="arya-practice-notes">Practice notes: confirm measurements and test the drivetrain before driving.</p>
```

## Popcorn Hack 3: Status Tones and Wide Layout

Use a wide action bar and status modifiers to distinguish a completed check from a task that needs attention.

```python
%%html
<!-- UI_RUNNER: Popcorn Hack 3 — Status Tones | interactive: true -->
<div class="ocs__links--wide">
  <button type="button" class="ocs__btn alert-green outline" onclick="document.getElementById('arya-button-status').textContent='Inspection passed.'">Mark Inspection Passed</button>
  <button type="button" class="ocs__btn alert-red fill" onclick="document.getElementById('arya-button-status').textContent='Issue flagged for the build team.'">Flag an Issue</button>
</div>
<p id="arya-button-status" role="status" aria-live="polite">No inspection result yet.</p>
```

## Homework Hack 1: Build Team Action Toolbar

Replace separate inline-styled actions with a shared OCS button group and utility modifiers. Each button updates the status rather than being decorative.

```python
%%html
<!-- UI_RUNNER: Homework Hack 1 — Build Team Toolbar | interactive: true -->
<div class="ocs__container"><div class="ocs__card" id="arya-build-toolbar">
  <h2 class="ocs__section-title">NYX Build Team Toolbar</h2>
  <p class="ocs__description">Choose the next task for the workshop.</p>
  <div class="ocs__links">
    <button type="button" class="ocs__btn alert-red fill pill" data-task="Fix the drivetrain alignment.">Needs Repair</button>
    <button type="button" class="ocs__btn alert-green fill pill" data-task="Inspection is complete and the robot is ready.">Ready for Testing</button>
    <button type="button" class="ocs__btn accent outline pill" data-task="Review the build log and team notes.">Review Notes</button>
  </div>
  <p class="ocs__callout" role="status" aria-live="polite" id="arya-build-result">Select a workshop action.</p>
</div></div>
<script>
(function () {
  const toolbar = document.getElementById('arya-build-toolbar');
  const result = document.getElementById('arya-build-result');
  toolbar.querySelectorAll('button[data-task]').forEach(function (button) {
    button.addEventListener('click', function () {
      result.textContent = button.dataset.task;
    });
  });
})();
</script>
```

## Homework Hack 2: Primary and Secondary Action Priority

Use a filled accent button for the main action, an outline for a secondary action, and an alert modifier for a warning. Button clicks change the visible check-in state.

```python
%%html
<!-- UI_RUNNER: Homework Hack 2 — Action Priority | interactive: true -->
<div class="ocs__container"><div class="ocs__card" id="arya-test-panel">
  <h2 class="ocs__section-title">Robot Test Check-In</h2>
  <p class="ocs__description">Record the outcome of a practice test.</p>
  <div class="ocs__links">
    <button type="button" class="ocs__btn accent fill" data-result="Test result recorded.">Record Test</button>
    <button type="button" class="ocs__btn outline" data-result="Showing latest test notes: check wheel alignment.">View Test Notes</button>
    <button type="button" class="ocs__btn alert-red outline" data-result="Test flagged for follow-up.">Report an Issue</button>
  </div>
  <output id="arya-test-result" role="status" aria-live="polite">No action selected.</output>
</div></div>
<script>
(function () {
  const panel = document.getElementById('arya-test-panel');
  const output = document.getElementById('arya-test-result');
  panel.querySelectorAll('button[data-result]').forEach(function (button) {
    button.addEventListener('click', function () {
      output.textContent = button.dataset.result;
    });
  });
})();
</script>
```

## Explanation

The `ocs__btn` base class supplies styling, while `fill`, `outline`, `pill`, and alert colors express action hierarchy. The OCS link groups handle layout without custom CSS. Buttons use click handlers and live feedback, so the actions are functional.
