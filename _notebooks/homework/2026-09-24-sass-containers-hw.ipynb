## Homework Hack: Robotics Workshop Hub

Use exactly one outer container and card, a three-column OCS grid with modifiers, a two-row table, and a callout.

```python
%%html
<!-- UI_RUNNER: container homework | interactive: true -->
<div class="ocs__container"><div class="ocs__card">
  <h2 class="ocs__section-title">NYX Workshop Hub</h2>
  <p class="ocs__description">Today's plan for team members.</p>
  <div class="ocs__grid ocs__grid--standard cols-3">
    <div class="ocs__grid-cell ocs__grid-cell--header">Workshop Details</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Start: 3:30 PM</div>
    <div class="ocs__grid-cell ocs__grid-cell--muted">Check-in: 3:15 PM</div>
  </div>
  <div class="ocs__table-wrap"><table class="ocs__table">
    <tr><th>Station</th><th>Status</th></tr>
    <tr><td>Drivetrain</td><td>Ready</td></tr>
    <tr><td>Intake</td><td>Parts review</td></tr>
  </table></div>
  <div class="ocs__callout">Remember to return shared tools after the session.</div>
</div></div>
```

## Explanation

The container centers content, the card groups the workshop, the grid adapts the details, and the table summarizes status.
