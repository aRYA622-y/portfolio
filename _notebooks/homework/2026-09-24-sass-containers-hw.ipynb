## Nonprofit Application

Poway Rotary uses OCS containers to present volunteer information for its community parade.

## Popcorn Hack: Parade Invite Rescue

One `ocs__container`, one `ocs__card`, and a three-column grid present the community parade invitation.

```python
%%html
<!-- UI_RUNNER: Rotary Parade Invite | interactive: true -->
<div class="ocs__container"><div class="ocs__card">
  <h2 class="ocs__section-title">Poway Rotary Parade Volunteer Invite</h2>
  <p class="ocs__description">Join Poway Rotary for parade-day community service.</p>
  <div class="ocs__grid ocs__grid--standard cols-3">
    <div class="ocs__grid-cell">Location: Poway Road</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Start: 9:30 AM</div>
    <div class="ocs__grid-cell">Activity: Parade support</div>
  </div>
</div></div>
```

## Homework Hack: Parade Hub Page

The hub uses one container and card, standard `cols-3` grid with required modifiers, an OCS table with two rows, and a concluding callout.

```python
%%html
<!-- UI_RUNNER: Rotary Parade Hub | interactive: true -->
<div class="ocs__container"><div class="ocs__card">
  <h2 class="ocs__section-title">Poway Rotary Community Parade</h2>
  <p class="ocs__description">Parade information for visitors and volunteers.</p>
  <div class="ocs__grid ocs__grid--standard cols-3">
    <div class="ocs__grid-cell ocs__grid-cell--header">Parade Day</div>
    <div class="ocs__grid-cell ocs__grid-cell--accent">Parade begins: 9:30 AM</div>
    <div class="ocs__grid-cell ocs__grid-cell--muted">Volunteer check-in: 8:30 AM</div>
  </div>
  <div class="ocs__table-wrap"><table class="ocs__table">
    <tr><th>Volunteer Station</th><th>Status</th></tr>
    <tr><td>Welcome Desk</td><td>Staffed</td></tr>
    <tr><td>Donation Collection</td><td>Accepting supplies</td></tr>
  </table></div>
  <div class="ocs__callout">Volunteers should check in before their assigned shift.</div>
</div></div>
```
