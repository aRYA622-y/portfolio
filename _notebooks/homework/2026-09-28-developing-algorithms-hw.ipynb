## Popcorn Hack 1: Translate Steps

Initialize an accumulator, traverse the input, and print the final sum.

```python
# CODE_RUNNER: popcorn 1
workshop_hours = [2, 4, 3, 5]
total = 0
for hours in workshop_hours:
    total += hours
print("Volunteer hours:", total)
```

## Popcorn Hack 2: Boundary Condition

Exactly the capacity is full, so use greater-than-or-equal instead of greater-than.

```python
# CODE_RUNNER: popcorn 2
signups = [8, 12, 15, 6]
full_count = 0
for number in signups:
    if number >= 12:
        full_count += 1
print("Full workshops:", full_count)
assert full_count == 2
```

## Popcorn Hack 3: Collector

Adapt the counter into a list of workshop names.

```python
# CODE_RUNNER: popcorn 3
names = ["CAD", "Build", "Coding", "Design"]
signups = [8, 12, 15, 6]
full = []
for name, count in zip(names, signups):
    if count >= 12:
        full.append(name)
print(full)
```

## Homework Hack: Inventory Restock Report

Decompose the task into ordered steps: traverse parallel lists, compare stock to a threshold, collect low items, and calculate required quantities.

```python
# CODE_RUNNER: homework - develop an algorithm
parts = ["Bolts", "Bearings", "Washers", "Spacers", "Wheels"]
stock = [12, 3, 8, 2, 6]
minimum = 6
restock = []
total_needed = 0
for index in range(len(parts)):
    if stock[index] < minimum:
        needed = minimum - stock[index]
        restock.append((parts[index], needed))
        total_needed += needed
print("Restock list:", restock)
print("Total units required:", total_needed)
assert restock == [("Bearings", 3), ("Spacers", 4)]
assert total_needed == 7
```

## Explanation

The algorithm processes each part once. Collecting names and quantities is more useful than only counting low-stock entries.
