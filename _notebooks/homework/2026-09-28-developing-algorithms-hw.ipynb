## Homework

```python
# CODE_RUNNER: Safe Passage restock report
items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]
target = [30, 20, 25, 15, 30]
def restock_report(items, in_stock, target):
    # Collect only items strictly below target, preserving corresponding shortages.
    short_items, short_amounts = [], []
    for i in range(len(items)):
        if in_stock[i] < target[i]:
            short_items.append(items[i])
            short_amounts.append(target[i] - in_stock[i])
    # Sum shortages using a loop.
    total_needed = 0
    for amount in short_amounts:
        total_needed += amount
    # Handle the empty case before trying to find a maximum.
    if len(short_items) == 0:
        print("Fully stocked!")
        return
    # Track the maximum shortage's position so its item can be reported.
    largest = 0
    for i in range(1, len(short_amounts)):
        if short_amounts[i] > short_amounts[largest]:
            largest = i
    # Combine the collected items, total, and maximum into one report.
    print("Restock needed:")
    for i in range(len(short_items)):
        print(f"{short_items[i]}: need {short_amounts[i]}")
    print("Total items needed:", total_needed)
    print(f"Most urgent: {short_items[largest]} (short by {short_amounts[largest]})")
restock_report(items, in_stock, target)
```

## Boundary test
At the target, Socks is not short. A fully stocked test should print only Fully stocked!

```python
# CODE_RUNNER: Fully stocked edge case
# Rerun the self-contained report definition in the homework cell first.
restock_report(["Socks"], [25], [25])
```
