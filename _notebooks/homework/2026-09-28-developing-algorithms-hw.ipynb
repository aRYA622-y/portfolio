## Nonprofit Application

Safe Passage Heals can combine smaller algorithms to identify shortages, total needed supplies, and find the most urgent restock item.

## Popcorn Hack 1: Translate Pseudocode

An accumulator traverses volunteer hours and computes an average.

```python
# CODE_RUNNER: algorithm popcorn 1
volunteer_hours = [3, 5, 2, 4]
total_hours = 0
for hours in volunteer_hours:
    total_hours += hours
average_hours = total_hours / len(volunteer_hours)
print("Total:", total_hours, "Average:", average_hours)
assert (total_hours, average_hours) == (14, 3.5)
```

## Popcorn Hack 2: Fix the Boundary

A workshop is full when signups meet or exceed capacity, including exact equality.

```python
# CODE_RUNNER: algorithm popcorn 2
signups = [12, 7, 15, 9]
capacity = 12
full_count = 0
for count in signups:
    if count >= capacity:
        full_count += 1
print("Full workshops:", full_count)
assert full_count == 2
```

## Popcorn Hack 3: Change Counter to Collector

Collect the names of workshops below minimum signup instead of only counting them.

```python
# CODE_RUNNER: algorithm popcorn 3
workshops = ["Resource Training", "Health Outreach", "Food Distribution", "Volunteer Orientation"]
signups = [12, 7, 15, 9]
below_target = []
for index in range(len(workshops)):
    if signups[index] < 12:
        below_target.append(workshops[index])
print(below_target)
assert below_target == ["Health Outreach", "Volunteer Orientation"]
```

## Homework Hack: Safe Passage Heals Supply Restock Report

Collect every shortage and its quantity, calculate the total using a loop, and identify the maximum shortage. Items at or above target are excluded.

```python
# CODE_RUNNER: developing algorithms nonprofit homework
items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [38, 7, 27, 6, 32]
target = [30, 20, 25, 15, 30]
short_items = []
short_amounts = []
for i in range(len(items)):
    shortage = target[i] - in_stock[i]
    if shortage > 0:
        short_items.append(items[i])
        short_amounts.append(shortage)
total_needed = 0
for amount in short_amounts:
    total_needed += amount
if short_amounts:
    largest_index = 0
    for i in range(1, len(short_amounts)):
        if short_amounts[i] > short_amounts[largest_index]:
            largest_index = i
    most_urgent = (short_items[largest_index], short_amounts[largest_index])
else:
    most_urgent = None
for name, amount in zip(short_items, short_amounts):
    print(f"{name}: need {amount}")
print("Total needed:", total_needed, "Most urgent:", most_urgent)
assert (short_items, short_amounts, total_needed, most_urgent) == (
    ["Blankets", "Shampoo"], [13, 9], 22, ("Blankets", 13))
```

## Explanation

The report uses three smaller algorithms: filter low inventory, total the shortages, and track the largest shortage with its matching item.
