## Popcorn
Prediction: changing Python index 1 adds 3 to the second level and appending 5 creates a fourth level: [6, 12, 12, 5], length 4. AP uses index 2 for that same second level.

```python
# CODE_RUNNER: UESL collection checkpoint
level_stars = [6, 9, 12]
level_stars[1] += 3
level_stars.append(5)
print(level_stars)
print(len(level_stars))
```

## MCQ
Answers: B, A, C, B. AP starts at 1; replacement preserves length; append adds one item; a list reduces maintenance rather than guaranteeing faster execution. Answer review: 4/4.

## Homework
Prediction: [12, 23, 10, 7], four levels, total 52.

```python
# CODE_RUNNER: UESL level collection
level_stars = [12, 18, 10]
if len(level_stars) >= 2:
    level_stars[1] += 5
level_stars.append(7)
total = 0
for stars in level_stars:
    total += stars
print(level_stars)
print(len(level_stars))
print(total)
assert (level_stars, len(level_stars), total) == ([12, 23, 10, 7], 4, 52)
```

## Tests
The empty test runs the total-only code, without appending a new level. Prediction: 0 levels, 0 stars.

```python
# CODE_RUNNER: UESL empty total test
level_stars = []
if len(level_stars) >= 2:
    level_stars[1] += 5
total = 0
for stars in level_stars:
    total += stars
print(len(level_stars))
print(total)
assert (len(level_stars), total) == (0, 0)
```

## Design Thinking
A game maker needs to add levels without rewriting the sum. A list plus traversal was chosen over 100 individual variables because new levels use append and the same loop. The empty test verified that the length guard avoids invalid indexing. Clear labels such as “Level 1” would let players read progress without understanding indices; a previously stored total still needs recalculation after a list changes.
