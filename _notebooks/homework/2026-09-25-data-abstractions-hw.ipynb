## Popcorn Hack 1: Inspect Scores

Read the first two items and the list length using Python's zero-based indexing.

```python
# CODE_RUNNER: popcorn 1
match_points = [8, 14, 11]
print(match_points[0], match_points[1], len(match_points))
```

## Popcorn Hack 2: Change the Collection

Update one score and append another. Replacing an item does not change list length; appending does.

```python
# CODE_RUNNER: popcorn 2
match_points = [8, 14, 11]
match_points[1] += 3
match_points.append(9)
print(match_points, len(match_points))
```

## Homework Hack: Tournament Score Collection

Guard against missing items, add a new result, and traverse the list with an accumulator instead of hardcoding the total.

```python
# CODE_RUNNER: homework - data abstraction
def tournament_report(scores):
    results = scores.copy()
    if len(results) > 1:
        results[1] += 2
    results.append(7)
    total = 0
    for score in results:
        total += score
    return results, len(results), total

print(tournament_report([10, 12, 9]))
assert tournament_report([10, 12, 9]) == ([10, 14, 9, 7], 4, 40)
assert tournament_report([]) == ([7], 1, 7)
```

## Design Thinking

The list represents ordered match scores. One collection is easier to extend and total than separate variables, and the length guard prevents an invalid index.
