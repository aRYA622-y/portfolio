## Nonprofit Application

An educational nonprofit can use a list to represent the points learners collect across successive activities in a community learning game.

## Popcorn Hack 1: Find a Level

Python indexing starts at zero. Read the first two scores and the length of the collection.

```python
# CODE_RUNNER: data abstraction popcorn 1
lesson_points = [14, 20, 15]
print(lesson_points[0], lesson_points[1], len(lesson_points))
assert (lesson_points[0], lesson_points[1], len(lesson_points)) == (14, 20, 3)
```

## Popcorn Hack 2: Update and Append

Change an existing item, then add a fourth lesson with append.

```python
# CODE_RUNNER: data abstraction popcorn 2
lesson_points = [6, 9, 12]
lesson_points[1] += 3
lesson_points.append(5)
print(lesson_points, len(lesson_points))
assert lesson_points == [6, 12, 12, 5]
```

## Homework Hack: Grow the Learning Collection

A list stores the ordered activity points. Check length before updating the second activity; use an accumulator to total all scores.

```python
# CODE_RUNNER: nonprofit learning activity collection
def activity_report(points):
    updated = list(points)
    if len(updated) >= 2:
        updated[1] += 5
    updated.append(7)
    total = 0
    for value in updated:
        total += value
    return updated, len(updated), total

print(activity_report([12, 18, 10]))
assert activity_report([12, 18, 10]) == ([12, 23, 10, 7], 4, 52)
assert activity_report([]) == ([7], 1, 7)
```

## Edge Case

An empty collection has a total of zero before the new activity is appended. Avoid indexing the missing second item.

```python
# CODE_RUNNER: check empty list without append
lesson_points = []
total_points = 0
if len(lesson_points) > 1:
    lesson_points[1] += 5
for value in lesson_points:
    total_points += value
assert (len(lesson_points), total_points) == (0, 0)
print("Activities:", len(lesson_points), "Total:", total_points)
```

## Design Thinking

One list replaces several separate variables and supports an arbitrary number of lessons. A length guard prevents index errors while iteration scales with the collection.
