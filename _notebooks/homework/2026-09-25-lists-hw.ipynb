## Popcorn Hack 3

```python
# CODE_RUNNER: Shoreline count and filter
items_requested = ["blankets", "socks", "meals", "socks", "water", "socks", "meals"]
frequency = {}
for item in items_requested:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1
popular = []
for item in frequency:
    if frequency[item] >= 2:
        popular.append(item)
print(frequency)
print(popular)
```

## Homework
Use explicit loops for totals, counting, and filtering. The average is converted to an integer as requested.

```python
# CODE_RUNNER: Shoreline supply drive report
donations_by_item = ["blankets", "socks", "blankets", "water", "socks", "blankets",
                     "meals", "socks", "blankets", "water", "blankets", "socks"]
costs = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]
print(len(donations_by_item))
print(donations_by_item[-1])
total = 0
for price in costs:
    total += price
print("Total:", int(total))
print("Average:", int(total / len(costs)))
frequency = {}
for item in donations_by_item:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1
popular = []
for item in frequency:
    if frequency[item] >= 3:
        popular.append(item)
print(frequency)
print(popular)
assert total == 31 and popular == ["blankets", "socks"]
```
