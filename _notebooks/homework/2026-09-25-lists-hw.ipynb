## Nonprofit Application

Safe Passage Heals can store requested supplies and associated costs in lists, then use indexing, accumulators, frequency counts, and filtering for its community distribution program.

## Popcorn Hack 1: Safe Indexing

Read the first and last item from a list of supplies without accessing an out-of-range index.

```python
# CODE_RUNNER: Safe Passage Heals lists popcorn 1
supplies = ["Soap", "Blankets", "Notebooks", "Socks"]
print("First:", supplies[0], "Last:", supplies[-1], "Count:", len(supplies))
assert (supplies[0], supplies[-1]) == ("Soap", "Socks")
```

## Popcorn Hack 2: Sum and Average

Loop through recorded unit costs to compute the total and average, including an empty-list safeguard.

```python
# CODE_RUNNER: lists popcorn 2
unit_costs = [7, 6, 8, 5, 7, 9]
total = 0
for cost in unit_costs:
    total += cost
average = total / len(unit_costs) if unit_costs else 0
print("Total:", total, "Average:", average)
assert (total, average) == (42, 7.0)
empty_costs = []
empty_average = sum(empty_costs) / len(empty_costs) if empty_costs else 0
assert empty_average == 0
```

## Popcorn Hack 3: Count and Filter

Count the frequency of each requested item and collect items requested at least twice.

```python
# CODE_RUNNER: lists popcorn 3
requests = ["Blankets", "Socks", "Blankets", "Soap", "Notebooks", "Socks", "Blankets"]
counts = {}
for item in requests:
    counts[item] = counts.get(item, 0) + 1
frequent = [item for item in counts if counts[item] >= 2]
print(counts, frequent)
assert set(frequent) == {"Blankets", "Socks"}
```

## Homework Hack: Distribution Request Report

Demonstrate first and last indexing, a cost accumulator, request counting, and a frequency-based filter.

```python
# CODE_RUNNER: Safe Passage Heals lists homework
requests = ["Soap", "Blankets", "Soap", "Socks", "Blankets", "Soap",
            "Notebooks", "Socks", "Soap", "Blankets", "Socks", "Soap"]
unit_costs = [2, 8, 2, 5, 8, 2, 3, 5, 2, 8, 5, 2]
first_request, last_request = requests[0], requests[-1]
total_cost = 0
for price in unit_costs:
    total_cost += price
average_cost = total_cost / len(unit_costs) if unit_costs else 0
frequency = {}
for item in requests:
    frequency[item] = frequency.get(item, 0) + 1
repeat_requests = [item for item in frequency if frequency[item] >= 2]
print("First:", first_request, "Last:", last_request)
print("Total:", total_cost, "Average:", average_cost)
print("Requested more than once:", repeat_requests)
assert first_request == last_request == "Soap"
assert total_cost == 52
assert set(repeat_requests) == {"Soap", "Blankets", "Socks"}
```

## Explanation

Lists store ordered requests and costs, while iteration calculates totals and a dictionary tracks frequency. Filtering identifies recurring supply needs.
