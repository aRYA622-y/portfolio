## Popcorn Hack 1: Safe Indexing

Access the first and last item without requesting an index outside the list.

```python
# CODE_RUNNER: popcorn 1
parts = ["Motor", "Wheel", "Sensor"]
print(parts[0], parts[-1], len(parts))
```

## Popcorn Hack 2: Accumulator and Average

Calculate a sum by traversing the list; guard against division by zero.

```python
# CODE_RUNNER: popcorn 2
measurements = [12, 16, 20, 24]
total = 0
for value in measurements:
    total += value
average = total / len(measurements) if measurements else 0
print(total, average)
```

## Popcorn Hack 3: Frequency and Filter

Count occurrences in a dictionary and keep entries that appear more than once.

```python
# CODE_RUNNER: popcorn 3
choices = ["Build", "Code", "Build", "Design", "Code", "Build"]
counts = {}
for choice in choices:
    counts[choice] = counts.get(choice, 0) + 1
popular = [name for name, count in counts.items() if count > 1]
print(counts, popular)
```

## Homework Hack: Robotics Station Survey

Combine safe indexing, numeric accumulation, frequency counting, and filtering in one report.

```python
# CODE_RUNNER: homework - lists
stations = ["Build", "Code", "Build", "CAD", "Build", "Code"]
hours = [2, 3, 2, 4, 3, 2]
first, last = stations[0], stations[-1]
total_hours = 0
for hour in hours:
    total_hours += hour
average_hours = total_hours / len(hours) if hours else 0
frequencies = {}
for station in stations:
    frequencies[station] = frequencies.get(station, 0) + 1
repeat_stations = [station for station in frequencies if frequencies[station] >= 2]
print(first, last, total_hours, average_hours)
print(frequencies, repeat_stations)
assert total_hours == 16 and set(repeat_stations) == {"Build", "Code"}
```

## Explanation

The list stores the sequence of responses. An accumulator totals hours, a dictionary counts responses, and filtering selects the repeated stations.
