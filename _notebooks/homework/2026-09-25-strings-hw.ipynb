## Nonprofit Application

Safe Passage Heals can use string methods to clean incoming outreach messages and format distribution status reports.

## Popcorn Hack 1: Extract the Message

Indexing and slicing recover known sections of a labeled outreach transmission.

```python
# CODE_RUNNER: strings popcorn 1
transmission = "xxDONATE---SUPPLIESxxREADY"
action = transmission[2:8]
item = transmission[11:19]
status = transmission[-5:]
print(action, item, status)
assert (action, item, status) == ("DONATE", "SUPPLIES", "READY")
```

## Popcorn Hack 2: Clean and Search

Trim spaces, strip extra punctuation, normalize case, and search for a delivery update.

```python
# CODE_RUNNER: strings popcorn 2
incoming = "  ...Volunteer>>>reports: SUPPLIES have ARRIVED!!!  "
clean = incoming.strip().replace("...", "").replace(">", "").replace("!", "")
clean = clean.lower()
print(clean)
print("Delivered:", "arrived" in clean)
print("Words:", len(clean.split()))
assert "arrived" in clean
```

## Popcorn Hack 3: Format an Outreach Report

Combine variables using concatenation, type conversion, join, and an f-string.

```python
# CODE_RUNNER: strings popcorn 3
volunteer = "Outreach Team"
delivered = 18
items = ["blankets", "socks", "notebooks"]
headline = volunteer + " delivered " + str(delivered) + " kits"
item_line = ", ".join(items)
report = f"{headline} | Items: {item_line}"
print(report)
assert "18 kits" in report
```

## Homework Hack: Decode an Outreach Update

Split a CSV-like string, use slicing and replacement, search the decoded text, and produce Markdown and JSON-style output.

```python
# CODE_RUNNER: strings nonprofit homework
raw_update = "  2026-09-29,OUTREACH,HYGIENE-KITS-DELIVERED,92  "
segments = [field.strip() for field in raw_update.strip().split(",")]
date, program, encoded_message, confidence = segments
decoded_message = encoded_message.replace("-", " ")
year = date[:4]
delivered = "delivered" in decoded_message.lower()
markdown_row = f"| {date} | {program} | {decoded_message} | {confidence}% |"
json_style = f'{{"date":"{date}","program":"{program}","message":"{decoded_message}","confidence":{confidence}}}'
print(markdown_row)
print(json_style)
assert year == "2026" and delivered and int(confidence) == 92
```

## Explanation

Slicing extracts known positions, replace and strip clean strings, and splitting supports field-based nonprofit reporting.
