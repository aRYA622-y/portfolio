## Popcorn Hack 1: Slicing

Use positive and negative slices to identify pieces of a message.

```python
# CODE_RUNNER: popcorn 1
message = "NYX-ROBOT-READY"
print(message[:3])
print(message[4:9])
print(message[-5:])
```

## Popcorn Hack 2: Clean a Message

Remove padding and punctuation, normalize the case, and count the words.

```python
# CODE_RUNNER: popcorn 2
raw = "  Robot, READY! Team ready.  "
clean = raw.strip().replace(",", "").replace("!", "").replace(".", "").lower()
print(clean)
print(clean.split().count("ready"))
```

## Popcorn Hack 3: Report Format

Combine individual pieces into a readable sentence with an f-string.

```python
# CODE_RUNNER: popcorn 3
team, result, location = "NYX", "passed", "practice field"
report = f"{team} inspection {result} at the {location}."
print(report)
```

## Homework Hack: Decode Team Update

Split structured text into fields, normalize the message, and produce a formatted report.

```python
# CODE_RUNNER: homework - strings
raw_update = " NYX | BUILD |  motors tested, drivetrain ready! "
team, section, update = [part.strip() for part in raw_update.split("|")]
clean_update = update.lower().replace("!", "").replace(",", "")
words = clean_update.split()
report = f"Team: {team}\nSection: {section.title()}\nUpdate: {clean_update}\nWords: {len(words)}"
print(report)
assert team == "NYX" and "drivetrain" in words and len(words) == 4
```

## Explanation

Slicing extracts known positions; string methods clean inconsistent input; splitting makes text searchable; formatting makes the results readable.
