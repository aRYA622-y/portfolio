## Homework
Prediction: original 40, stored 40 after the bonus, recalculated 45.

```python
# CODE_RUNNER: UESL star counter
game_name = "UESL Comet Quest"
reduced_motion = True
level_one_stars, level_two_stars, level_three_stars = 12, 18, 10
total_stars = level_one_stars + level_two_stars + level_three_stars
print(game_name)
print("Reduced motion:", reduced_motion)
print("Original:", total_stars)
level_two_stars += 5
print("Stored after bonus:", total_stars)
total_stars = level_one_stars + level_two_stars + level_three_stars
print("Recalculated:", total_stars)
```

## Tests
Fresh starting values for each case. Bonus 5 should produce 40/40/45; bonus 0 should produce 40/40/40.

```python
# CODE_RUNNER: UESL fresh bonus tests
for bonus in [5, 0]:
    first, second, third = 12, 18, 10
    original = first + second + third
    stored = original
    second += bonus
    updated = first + second + third
    print("Bonus:", bonus, "Totals:", original, stored, updated)
    assert (original, stored, updated) == (40, 40, 40 + bonus)
```

## Design Thinking
A UESL player may need less distracting motion. The goal is to make progress clear and let the player choose reduced motion; a Boolean represents that choice. This prototype uses separate integer counts and recalculates the total. Testing bonus 0 confirmed that recalculation preserves 40 when no stars change. With 100 levels, maintaining separate names and a long sum would be cumbersome; a list would simplify those edits.
