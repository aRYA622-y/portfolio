## Nonprofit Application

An educational nonprofit can use a simple accessible learning game to help participants practice skills. Variables track points and store reduced-motion preferences.

## Popcorn Hack: Save a Checkpoint

A stored integer keeps the earlier point total even if the live score changes.

```python
# CODE_RUNNER: learning-game checkpoint
points = 8
checkpoint_points = points
points = points + 4
print("Current points:", points)
print("Checkpoint points:", checkpoint_points)
assert (points, checkpoint_points) == (12, 8)
```

## Homework Hack: Accessible Learning Game

Use a string, Boolean setting, and separate numeric level scores. Recalculate the total after awarding a participation bonus.

```python
# CODE_RUNNER: nonprofit learning-game homework
game_name = "Community Learning Quest"
reduced_motion = True
level_one_points = 12
level_two_points = 16
level_three_points = 12
stored_total = level_one_points + level_two_points + level_three_points
bonus = 5
level_two_points += bonus
updated_total = level_one_points + level_two_points + level_three_points
print(game_name, "reduced motion:", reduced_motion)
print("Stored total:", stored_total, "Updated total:", updated_total)
assert (stored_total, updated_total) == (40, 45)
```

## Tests

Check that adding five points changes the recalculated total but not the stored total; adding zero leaves both totals equal.

```python
# CODE_RUNNER: homework tests
for bonus in [5, 0]:
    level_one, level_two, level_three = 12, 16, 12
    original_total = level_one + level_two + level_three
    level_two += bonus
    revised_total = level_one + level_two + level_three
    print("Bonus:", bonus, "Original:", original_total, "Revised:", revised_total)
    assert revised_total == 40 + bonus
```

## Design Thinking

A reduced-motion Boolean makes the educational game more accessible. Separate level variables demonstrate assignment and recalculation; a list would become useful as the program grows.
