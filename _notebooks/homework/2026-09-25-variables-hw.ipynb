## Homework Hack: Robot Match Tracker

Use integers, a string, and a Boolean; change a bonus and recalculate the final score using variables.

```python
# CODE_RUNNER: homework - variables and assignment
robot_name = "NYX Prototype"
autonomous_points = 12
teleop_points = 24
bonus_points = 5
inspection_passed = True
match_total = autonomous_points + teleop_points + bonus_points
print(robot_name, "score:", match_total)
bonus_points = 0
match_total = autonomous_points + teleop_points + bonus_points
print("Without bonus:", match_total)
print("Inspection:", inspection_passed)
assert match_total == 36
```

## Design Thinking

Separate values from calculations so match scoring can change without rewriting the program. The first score is 41 and the updated score is 36.
