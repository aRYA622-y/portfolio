## Popcorn Hack 1: Arithmetic

Compute four operator results from variables.

```python
# CODE_RUNNER: popcorn 1
first, second, third = 90, 15, 6
print("Sum:", first + second + third)
print("Difference:", first - second)
print("Product:", second * third)
print("Quotient:", first / second)
```

## Popcorn Hack 2: Workshop Seating

Calculate occupied seats, remaining seats, and seats per room.

```python
# CODE_RUNNER: popcorn 2
capacity = 96
teams = 5
students_per_team = 4
occupied = teams * students_per_team
remaining = capacity - occupied
print("Occupied:", occupied)
print("Remaining:", remaining)
print("Per room:", remaining / 2)
```

## Homework Hack: Club Supply Budget

Calculate weekly and monthly expenses, a one-time registration fee, and a daily average.

```python
# CODE_RUNNER: homework - arithmetic operators
cost_per_session = 28
sessions_per_week = 3
weeks = 4
registration_fee = 40
weekly = cost_per_session * sessions_per_week
monthly = weekly * weeks
first_month = monthly + registration_fee
average_daily = monthly / 28
print("Weekly:", weekly)
print("Monthly:", monthly)
print("First month:", first_month)
print("Daily average:", average_daily)
assert (weekly, monthly, first_month, average_daily) == (84, 336, 376, 12.0)
```
