## Popcorn Hack 1: Four Arithmetic Operators

Calculate addition, subtraction, multiplication, and division with stored variables.

```python
# CODE_RUNNER: arithmetic popcorn
value_a, value_b, value_c = 100, 10, 25
print("Total:", value_a + value_b + value_c)
print("Difference:", value_a - value_b)
print("Product:", value_b * value_c)
print("Division:", value_a / value_b)
assert (value_a + value_b + value_c, value_a - value_b, value_b * value_c, value_a / value_b) == (135, 90, 250, 10.0)
```

## Popcorn Hack 2: Recovery Center Booking

Find used spots, remaining spots, and the even split across two rooms.

```python
# CODE_RUNNER: booking popcorn
monthly_capacity = 100
spots_per_group = 4
sessions_booked = 3
occupied_spots = spots_per_group * sessions_booked
available_spots = monthly_capacity - occupied_spots
per_room = available_spots / 2
print("Occupied:", occupied_spots, "Remaining:", available_spots, "Per room:", per_room)
assert (occupied_spots, available_spots, per_room) == (12, 88, 44.0)
```

## Homework Hack: Recovery Center Cost Report

Calculate weekly cost, monthly cost, the first month including a one-time enrollment fee, and average daily monthly cost.

```python
# CODE_RUNNER: Poway Recovery Center homework
cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4
enrollment_fee = 75
weekly_cost = cost_per_session * sessions_per_week
monthly_cost = weekly_cost * weeks_per_month
first_month_cost = monthly_cost + enrollment_fee
average_daily_cost = monthly_cost / 30
print("Weekly:", weekly_cost, "Monthly:", monthly_cost)
print("First month:", first_month_cost, "Daily average:", average_daily_cost)
assert (weekly_cost, monthly_cost, first_month_cost, average_daily_cost) == (105, 420, 495, 14.0)
```

## Explanation

Multiplication gives combined session costs, subtraction finds remaining capacity, and division produces room allocations and daily averages.
