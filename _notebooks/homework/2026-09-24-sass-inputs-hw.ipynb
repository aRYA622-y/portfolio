## Homework Hack: Workshop Registration

Provide a search/category field, first-name field, and properly typed email field.

```python
%%html
<!-- UI_RUNNER: inputs homework | interactive: true -->
<div class="ocs__card">
  <h2 class="ocs__section-title">Workshop Registration</h2>
  <label>Workshop topic <input class="ocs__input large gradient" type="text" placeholder="Build, CAD, or coding"></label>
  <label>First name <input class="ocs__input small" type="text" placeholder="First name"></label>
  <label>Email <input class="ocs__input medium" type="email" placeholder="name@example.com"></label>
</div>
```

## Explanation

Input classes reuse the OCS styling; type=email supplies browser-level email-format support.
