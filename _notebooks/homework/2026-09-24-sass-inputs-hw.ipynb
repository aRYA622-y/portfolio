## Nonprofit Application

## Popcorn Hack: PVO Input Grammar

The resource search is large and gradient-styled; the volunteer first-name field is compact.

```python
%%html
<!-- UI_RUNNER: PVO Popcorn Inputs | interactive: true -->
<input class="ocs__input large gradient" type="search" placeholder="Search PVO resources...">
<input class="ocs__input small" type="text" placeholder="Volunteer first name">
```

## Homework Hack: PVO Volunteer Signup

Three fields use the shared OCS input class with appropriate modifiers and email input type.

```python
%%html
<!-- UI_RUNNER: PVO Volunteer Signup | interactive: true -->
<div class="ocs__card">
  <h2 class="ocs__section-title">PVO Volunteer Signup</h2>
  <label>Assistance category <input class="ocs__input large gradient" type="text" placeholder="Community outreach"></label>
  <label>First name <input class="ocs__input small" type="text" placeholder="Volunteer name"></label>
  <label>Email <input class="ocs__input medium" type="email" placeholder="volunteer@example.org"></label>
</div>
```
