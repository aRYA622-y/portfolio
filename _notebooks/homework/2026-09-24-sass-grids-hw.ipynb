## Popcorn Hack 2: Control Beaker Width

The final beaker receives `2fr`, twice the width of each other column.

```python
%%html
<!-- UI_RUNNER: Control Beaker | interactive: true -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns:1fr 1fr 1fr 2fr">
  <div class="ocs__grid-cell">Test 1</div>
  <div class="ocs__grid-cell">Test 2</div>
  <div class="ocs__grid-cell">Test 3</div>
  <div class="ocs__grid-cell">Control</div>
</div>
```
