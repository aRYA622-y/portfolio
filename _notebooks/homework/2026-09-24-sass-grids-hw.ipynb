## Nonprofit Application

A nonprofit community STEM workshop uses OCS grids to present science demonstrations and temperature results.

## Popcorn Hack 1: Beaker Grid

Use one standard OCS grid with three beaker cells.

```python
%%html
<!-- UI_RUNNER: Nonprofit STEM Beakers | interactive: true -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell">Beaker 1: Water</div>
  <div class="ocs__grid-cell">Beaker 2: Salt water</div>
  <div class="ocs__grid-cell">Beaker 3: Vinegar</div>
</div>
```

## Popcorn Hack 2: Control Beaker Width

The final beaker receives `2fr`, twice the width of each other column.

```python
%%html
<!-- UI_RUNNER: Control Beaker | interactive: true -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns:1fr 1fr 1fr 2fr">
  <div class="ocs__grid-cell">Test 1</div>
  <div class="ocs__grid-cell">Test 2</div>
  <div class="ocs__grid-cell">Test 3</div>
  <div class="ocs__grid-cell">Control</div>
</div>
```

## MCQ Check

`cols-4` creates four columns, `2fr 1fr` yields a 2:1 ratio, `ocs__grid-cell--header` spans a row, and custom `display:flex` breaks the OCS-only rule.

## Homework Hack: Temperature Grid and Calculator
