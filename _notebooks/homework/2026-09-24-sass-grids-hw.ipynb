## Homework Hack: Trial Results and Keypad

Use a header-spanning standard grid and an accent cell for the outlier; create a separate calculator grid.

```python
%%html
<!-- UI_RUNNER: grid homework | interactive: true -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell ocs__grid-cell--header">Drivetrain Trial (cm)</div>
  <div class="ocs__grid-cell">Trial 1: 45</div>
  <div class="ocs__grid-cell">Trial 2: 47</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Trial 3: 61 (outlier)</div>
</div>
<br>
<div class="ocs__grid ocs__grid--calculator">
  <div class="ocs__grid-cell">7</div><div class="ocs__grid-cell">8</div>
  <div class="ocs__grid-cell">9</div><div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
</div>
```

## Explanation

The header modifier spans the row. An accent highlights the unusual result without changing the other cells.
