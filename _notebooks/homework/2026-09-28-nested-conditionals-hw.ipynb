## Popcorn Hack 1
For True/False, the outer weekday branch is taken, then its homework-finished else: Finish homework. Test all three outcomes.

````python
# CODE_RUNNER: After-school plan paths
for is_weekday, homework_finished in [(True,False),(True,True),(False,False)]:
    if is_weekday:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
    else:
        print("Weekend plan")
````

## Popcorn Hack 2
The two original if statements were siblings, so permission could be checked even without login. Indenting the second decision makes it depend on login.

````python
# CODE_RUNNER: Repair student account nesting
for logged_in, has_permission in [(True,True),(True,False),(False,True)]:
    if logged_in:
        print("Account found")
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")
````

## Popcorn Hack 3

````python
# CODE_RUNNER: Four lunch and practice paths
for lunch_packed, practice_after_school in [(True,True),(True,False),(False,True),(False,False)]:
    if lunch_packed:
        if practice_after_school:
            print("Bring packed lunch and an extra practice snack")
        else:
            print("Bring packed lunch")
    else:
        if practice_after_school:
            print("Buy lunch and bring a practice snack")
        else:
            print("Buy lunch")
````

## Homework
The original starter values are preserved; additional cases exercise every meaningful branch.

````python
# CODE_RUNNER: Daily routine dashboard
student = {"logged_in":True,"homework_finished":False,"lunch_packed":True,"practice_today":True}
def daily_dashboard(student):
    if student["logged_in"]:
        if student["homework_finished"]:
            print("Homework complete: free time before your next activity")
        else:
            print("Finish homework before taking free time")
        if student["lunch_packed"]:
            if student["practice_today"]:
                print("Packed lunch: add a practice snack")
            else:
                print("Bring your packed lunch")
        else:
            if student["practice_today"]:
                print("Buy lunch and bring a practice snack")
            else:
                print("Buy lunch at school")
    else:
        print("Log in first; schedule unavailable")
daily_dashboard(student)
print("All meaningful paths:")
daily_dashboard({"logged_in":False,"homework_finished":False,"lunch_packed":False,"practice_today":False})
for homework_finished in [False,True]:
    for lunch_packed in [False,True]:
        for practice_today in [False,True]:
            daily_dashboard({"logged_in":True,"homework_finished":homework_finished,"lunch_packed":lunch_packed,"practice_today":practice_today})
````

## Explanation
The account check controls whether schedule recommendations can appear. Inside the authorized branch, lunch decides which practice-related recommendation is appropriate. The tests cover both homework outcomes, all four lunch/practice combinations, and the logged-out path.
