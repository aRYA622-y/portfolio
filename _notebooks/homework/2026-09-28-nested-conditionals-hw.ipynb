## Popcorn Hack 1: Homework Plan

Choose a study plan based on the amount of work and whether help is available.

```python
# CODE_RUNNER: popcorn 1
for hours, tutor in [(3, True), (3, False), (1, True)]:
    if hours >= 2:
        if tutor:
            print("Work through difficult problems with a tutor")
        else:
            print("Start with the longest assignment")
    else:
        print("Finish the short assignment")
```

## Popcorn Hack 2: Account Check

An active account must also have a valid student ID.

```python
# CODE_RUNNER: popcorn 2
for active, valid_id in [(True, True), (True, False), (False, True)]:
    if active:
        if valid_id:
            print("Access granted")
        else:
            print("Verify student ID")
    else:
        print("Account inactive")
```

## Popcorn Hack 3: Lunch and Practice

Trace combinations of the two conditions.

```python
# CODE_RUNNER: popcorn 3
for ate_lunch, has_practice in [(True, True), (True, False), (False, True), (False, False)]:
    if ate_lunch:
        if has_practice:
            print("Pack water for practice")
        else:
            print("Study after lunch")
    else:
        if has_practice:
            print("Eat before practice")
        else:
            print("Take a lunch break")
```

## Homework Hack: School Day Dashboard

Use nested decisions for account access, homework priority, and tennis practice.

```python
# CODE_RUNNER: homework - nested conditionals
def day_plan(student):
    if not student["account_active"]:
        return "Activate the school account first"
    if not student["id_valid"]:
        return "Verify the student ID first"
    if student["homework_hours"] >= 2:
        if student["practice"]:
            return "Start homework early before tennis"
        return "Set aside a focused homework block"
    if student["practice"]:
        return "Review notes, then go to practice"
    return "Complete the short assignment"

cases = [
 {"account_active": False, "id_valid": True, "homework_hours": 2, "practice": True},
 {"account_active": True, "id_valid": False, "homework_hours": 1, "practice": False},
 {"account_active": True, "id_valid": True, "homework_hours": 3, "practice": True},
 {"account_active": True, "id_valid": True, "homework_hours": 1, "practice": False}]
for case in cases:
    print(day_plan(case))
assert day_plan(cases[2]) == "Start homework early before tennis"
```

## Explanation

Nested conditionals check the prerequisite first and the more specific choice second. Different test records exercise the main branches.
