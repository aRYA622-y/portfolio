## Nonprofit Application

PVO can use nested decisions to manage volunteer access, open shifts, meal distribution, and delivery schedules.

## Popcorn Hack 1: Volunteer Shift Plan

The outer branch checks whether it is an outreach day; the inner branch checks whether a volunteer has finished preparation.

````python
# CODE_RUNNER: nested popcorn 1
for outreach_day, prepared in [(True, True), (True, False), (False, False)]:
    if outreach_day:
        if prepared:
            print("Ready for the outreach shift")
        else:
            print("Finish preparing supplies")
    else:
        print("Review next outreach date")
````

## Popcorn Hack 2: Repair Volunteer Account Check

Permission is checked only after successful login.

````python
# CODE_RUNNER: nested popcorn 2
for logged_in, permission in [(True, True), (True, False), (False, True)]:
    if logged_in:
        if permission:
            print("Volunteer dashboard available")
        else:
            print("Access request needed")
    else:
        print("Sign in before accessing volunteer data")
````

## Popcorn Hack 3: Meal Distribution and Delivery

Trace all four combinations of packed meals and scheduled delivery.

````python
# CODE_RUNNER: nested popcorn 3
for meals_packed, delivery_today in [(True, True), (True, False), (False, True), (False, False)]:
    if meals_packed:
        if delivery_today:
            print("Load packed meals for delivery")
        else:
            print("Store packed meals safely")
    else:
        if delivery_today:
            print("Prepare meals before loading")
        else:
            print("Review the next meal preparation shift")
````

## Homework Hack: PVO Volunteer Dashboard

Only an authorized volunteer can view the schedule. Nested decisions then select work priorities based on open shift hours and delivery demand.

````python
# CODE_RUNNER: PVO nested-conditionals homework
def volunteer_dashboard(account):
    if account["logged_in"]:
        if account["authorized"]:
            if account["open_shift_hours"] >= 2:
                if account["delivery_today"]:
                    return "Prioritize filling delivery shifts"
                return "Recruit volunteers for open shifts"
            if account["meals_packed"]:
                return "Confirm the next distribution appointment"
            return "Prepare meal kits for distribution"
        return "Permission required"
    return "Please sign in"

tests = [
 {"logged_in":False,"authorized":True,"open_shift_hours":3,"delivery_today":True,"meals_packed":True},
 {"logged_in":True,"authorized":False,"open_shift_hours":1,"delivery_today":False,"meals_packed":True},
 {"logged_in":True,"authorized":True,"open_shift_hours":3,"delivery_today":True,"meals_packed":True},
 {"logged_in":True,"authorized":True,"open_shift_hours":0,"delivery_today":False,"meals_packed":False}
]
for case in tests:
    print(volunteer_dashboard(case))
assert volunteer_dashboard(tests[2]) == "Prioritize filling delivery shifts"
````

## Explanation

Access is checked before private shift information. Within the authorized branch, delivery urgency refines the action for open shifts.
